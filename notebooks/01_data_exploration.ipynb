```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load the dataset
df = pd.read_csv('../data/dataset.csv')  # adjust filename if different

# First look
print(df.shape)        # how many rows and columns?
df.head()              # see first 5 rows


# What columns do we have?
df.columns.tolist()

# Check for missing values
df.isnull().sum()

# Get a feel for the audio features
df[['energy', 'valence', 'danceability', 'tempo', 'acousticness']].describe()

# Visualize distributions
features = ['energy', 'valence', 'danceability', 'tempo', 'acousticness']
df[features].hist(bins=30, figsize=(12, 8))
plt.tight_layout()
plt.show()
```
